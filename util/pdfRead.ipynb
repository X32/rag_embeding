```python
import pdfplumber

with pdfplumber.open("../data/README.pdf") as pdf:
    print(len(pdf.pages))
    page = pdf.pages[0]  # 获取第一页
    table = page.extract_table()  # 提取表格
    for row in table:
        print(row)
    text = page.extract_text()
    if text:
        print(text)
```

```python
with pdfplumber.open("../data/README.pdf") as pdf:
    all_text = []
    for page in pdf.pages:  # 遍历所有页面
        text = page.extract_text()
        if text:  # 避免空页面
            all_text.append(text)
    # 合并所有页面文本
    full_text = "\n".join(all_text)
    print(full_text)
```

```python
from pdf2image import convert_from_path

# 处理单页 PDF，默认只取第 1 页
pages = convert_from_path('../data/cert1.pdf', dpi=500)
```

```python
print(len(pages))
```

```python
page = pages[0]  # 单页PDF，取第1个元素（唯一的page）

# 2. 显示图片（弹出系统默认查看器）
page.show()
```

```python
import numpy as np
page_np = np.array(page.convert("RGB"))
```

```python

def get_ocr(use_cuda: bool = True) -> "RapidOCR":
    try:
        from rapidocr_paddle import RapidOCR
        ocr = RapidOCR(det_use_cuda=use_cuda, cls_use_cuda=use_cuda, rec_use_cuda=use_cuda)
    except ImportError:
        from rapidocr_onnxruntime import RapidOCR
        ocr = RapidOCR()
    return ocr

ocr = get_ocr()
result, _ = ocr(page_np)
```

```python
resp = ''
if result:
    ocr_result = [line[1] for line in result]
    for res in ocr_result:
        resp+=res+'\n'
    # resp += " ".join(ocr_result)
print(resp)
```
